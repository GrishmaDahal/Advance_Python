**Question 6: List Sanitization and Metrics**

Given:

marks = [45, -3, 88, "NA", 102, None, 67]
Write a function that:
1. Removes invalid entries
2. Returns:

    Cleaned list

    Number of invalid entries

    Percentage of valid marks
    
Return values using a tuple.

```python
def sanitize_marks(marks):
    """
    Cleans a list by removing non-numeric values and marks outside 0-100.
    Returns: (cleaned_list, invalid_count, valid_percentage)
    """
    
    tot_entries = len(marks)  # Total number of original items
    cleaned = []              # List to store only valid marks
    
    for mark in marks:
        
        # isinstance(mark, (int, float)) ensures we only keep numbers
        # 0 <= mark <= 100 ensures the score is within a logical percentage range
        if isinstance(mark, (int, float)) and 0 <= mark <= 100: 
            cleaned.append(mark)

    invalid_cnt = tot_entries - len(cleaned)
    
    # Round to 2 decimal places for cleaner output
    valid_per = round((len(cleaned) / tot_entries) * 100, 2)

    return (cleaned, invalid_cnt, valid_per)

marks = [45, -3, 88, "NA", 102, None, 67]
result= sanitize_marks(marks)
cleaned_list, invalid_count, valid_percentage = result

print("\n--- Sanitization Metrics ---")
print(f"Cleaned List of Valid Marks: {cleaned_list}")
print(f"Total Invalid Entries Found: {invalid_count}")
print(f"Percentage of Valid Marks: {valid_percentage}%")
    
    
```

**Question 8: Prime Number Function**

Write is_prime(n) and display prime numbers between 1 and 50.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
        return True

prime_number = []

for i in range(1, 51):
    if is_prime(i):
        prime_number.append(i)
print("Prime Numbers between 1 and 50:")
print(prime_number)
```

**Question 9: Function Composition**

Write functions to:
1. Calculate total marks
2. Calculate average
3. Classify result (Distinction / Pass / Fail)

Then write a function that combines them and returns:

(name, total, average, result)

Explain why modular design is beneficial.

```python
def total_marks(marks):
    return sum(marks)


def average_marks(marks):
    return sum(marks) / len(marks)


def classify(avg):
    if avg >= 75:
        return "Distinction"
    elif avg >= 40:
        return "Pass"
    else:
        return "Fail"

def student_result(name, marks):
    total = total_marks(marks)
    avg = average_marks(marks)
    result = classify(avg)
    return name, total, avg, result

# Testing the function with different student data
print(student_result("Aman", [65, 80, 90]))
print(student_result("Daman", [80, 45, 50]))
print(student_result("Saman", [15, 50, 30]))

# Modular design improves readability, testing, and reuse of code.
```

**Question 10: Defensive Input Parsing**

Write a function that accepts a raw input string in the format:

"Ram|101|22|78, 88, 91"

Hint: Parse student data from string format: "Name|Roll|Age|Mark1, Mark2, Mark3"

Tasks:
1. Extract name, roll number, age, and marks
2. Convert them into appropriate data types
3. Validate age (≥16) and marks (0–100)
4. Return the result as a dictionary using roll number as the key

The function must not crash for malformed input.

```python
def parse_student_data(raw_input):
    """
    Parses a string in 'Name|Roll|Age|Mark1, Mark2, Mark3' format.
    Returns a dictionary with Roll Number as the key if valid, or an error message.
    """
    try:
        # Extract components using delimiters
        # We split by '|' first, ensuring we have exactly 4 parts
        parts = [p.strip() for p in raw_input.split('|')]
        if len(parts) != 4:
            return {"error": "Malformed input: Expected 4 segments separated by '|'"}

        name, roll, age_str, marks_raw = parts

        # Task 2: Convert to appropriate data types
        roll_id = int(roll)
        age = int(age_str)
        # Convert comma-separated marks into a list of integers
        marks = [int(m.strip()) for m in marks_raw.split(',')]

        # Task 3: Validate age (≥16) and marks (0–100)
        if age < 16:
            return {"error": f"Validation failed: Age {age} is below the minimum requirement (16)"}
        
        for m in marks:
            if not (0 <= m <= 100):
                return {"error": f"Validation failed: Mark {m} is out of the valid range (0-100)"}

        # Task 4: Return as dictionary with roll number as the key
        return {
            roll_id: {
                "name": name,
                "age": age,
                "marks": marks
            }
        }
        
    except ValueError:
        # Handles cases where non-numeric strings are provided for Roll, Age, or Marks
        return {"error": "Data conversion error: Non-numeric value found in numeric field"}
    except Exception as e:
        # Catch-all for any other unexpected issues to prevent crashing
        return {"error": f"Unexpected error: {str(e)}"}

# Test cases
print(parse_student_data("Ram|101|22|78, 88, 91"))    # Valid input
print(parse_student_data("Shyam|102|15|90, 85"))      # Invalid age
print(parse_student_data("Invalid_String"))           # Malformed input
```

**Question 11: Type Stability and Conversion**
Given the list:

data = [10, "20", 30.5, "abc", None, True]

Write a function that:
1. Separates numeric and non-numeric values
2. Converts numeric values to float
3. Returns both lists

Explain in comments why True behaves as a numeric value.

```python
def separate_and_convert(data_list):
    numeric_values = []
    non_numeric_values = []
    
    for item in data_list:
        # Check if the item is an int, float, or bool (True/False)
        # Note: None must be explicitly excluded as it is not numeric
        if isinstance(item, (int, float, bool)) and item is not None:
            # Convert numeric values to float
            numeric_values.append(float(item))
        else:
            non_numeric_values.append(item)
            
    return numeric_values, non_numeric_values

# Given list
data = [10, "20", 30.5, "abc", None, True]

# Processing
numeric, non_numeric = separate_and_convert(data)

# Output
print(f"Numeric List (as floats): {numeric}")
print(f"Non-Numeric List: {non_numeric}")

"""
Explanation :
In Python, 'True' behaves as a numeric value because the 'bool' class is a 
subclass of the 'int' class. Internally, 'True' is represented as the integer 1 
and 'False' as the integer 0. This design was maintained for backward 
compatibility, allowing booleans to be used in mathematical operations 
(e.g., True + True = 2).
"""
```

**Question 12: Pattern-Based Looping**

Generate the sequence:

2, 6, 12, 20, 30, 42, 56

Rules:

Do not hardcode the list

Use loops and conditionals only

Explain the pattern in comments

```python
def generate_sequence():
    """
    Generates the sequence 2, 6, 12, 20, 30, 42, 56 
    using loops and the pattern n*(n+1).
    """
    sequence = []
    # The sequence has 7 elements
        # The pattern for the nth term is n * (n + 1)
        # Term 1: 1 * 2 = 2
        # Term 2: 2 * 3 = 6
        # ... and so on
    for n in range(1, 8):
        term = n * (n + 1)
        sequence.append(term)
        
    return sequence

# Generate the sequence
result = generate_sequence()
print(f"Generated Sequence: {result}")

"""
Explanation :
The pattern can be identified in two ways:
1. The difference between consecutive terms increases by 2 each time (4, 6, 8, 10, 12, 14).
2. The nth term in the sequence is calculated by multiplying n by (n + 1). 
   This is a sequence of pronic numbers or oblong numbers.
"""
```

**Question 13: Set-Based Enrollment Analysis**

Given:

python = {"Amit", "Sita", "Hari", "Nita"}

sql = {"Hari", "Ram", "Nita"}

statistics = {"Amit", "Hari", "Gita"}

Find:
1. Students enrolled in exactly two subjects
2. Students enrolled in only one subject
3. Students enrolled in all subjects

```python
# Initial Data
python = {"Amit", "Sita", "Hari", "Nita"}
sql = {"Hari", "Ram", "Nita"}
statistics = {"Amit", "Hari", "Gita"}

# 1. Students enrolled in all subjects
# Intersection of all three sets
all_three = python & sql & statistics

# 2. Students enrolled in exactly two subjects
# Strategy: Find intersections of pairs and subtract those in all three
p_s = (python & sql) - all_three
s_st = (sql & statistics) - all_three
st_p = (statistics & python) - all_three
exactly_two = p_s | s_st | st_p

# 3. Students enrolled in only one subject
# Strategy: Total students minus those in two or three subjects
all_students = python | sql | statistics
more_than_one = exactly_two | all_three
only_one = all_students - more_than_one

# Results
print(f"Enrolled in all subjects: {all_three}")
print(f"Enrolled in exactly two: {exactly_two}")
print(f"Enrolled in only one: {only_one}")
```

**Question 14: Dictionary Aggregation Logic**

Given:

records = {

    "Amit": [78, 82, 91],
    "Sita": [88, 79, 85],
    "Hari": [45, 52, 60]
}

Write a function that:
1. Computes average per student
2. Identifies top and bottom performers
3. Returns results using a single dictionary

```python
def aggregate_student_performance(records):
    """
    Computes averages per student and identifies top/bottom performers.
    Returns results in a single aggregated dictionary.
    """
    # 1. Compute average per student
    # We use a dictionary comprehension to build a new dictionary of averages
    averages = {name: sum(marks) / len(marks) for name, marks in records.items()}
    
    # 2. Identify top and bottom performers
    # max() and min() can take a 'key' argument to compare dictionary keys 
    # based on their associated values (the averages we just calculated).
    top_performer = max(averages, key=averages.get)
    bottom_performer = min(averages, key=averages.get)
    
    # 3. Return results in a single dictionary
    return {
        "student_averages": averages,
        "top_performer": top_performer,
        "bottom_performer": bottom_performer
    }

# Given data
records = {
    "Amit": [78, 82, 91],
    "Sita": [88, 79, 85],
    "Hari": [45, 52, 60]
}

# Execute and display result
performance_summary = aggregate_student_performance(records)
print(f"Performance Summary: {performance_summary}")
```

**Question 15: Nested Data Structure Analysis**

Given a nested dictionary representing a school's grade book:

school_data = {

    "Class A": {
        "Amit": [78, 82, 91],
        "Sita": [88, 79, 85]
    },
    "Class B": {
        "Hari": [45, 52, 60],
        "Ram": [92, 88, 95]
    }
}
**Write a function that:**

1. Calculates the average marks for each student
2. Finds the top performer in each class
3. Calculates the overall school average
4. Returns a dictionary with:

class_toppers: {class_name: (student_name, average)}

school_average: float

student_averages: {student_name: average}

**Explain in comments why nested dictionaries are useful for hierarchical data.**

```python
school_data = {
    "Class A": {
        "Amit": [78, 82, 91],
        "Sita": [88, 79, 85]
    },
    "Class B": {
        "Hari": [45, 52, 60],
        "Ram": [92, 88, 95]
    }
}

def analyze(data):

    class_toppers = {}      # To store {Class: (Name, Avg)}
    student_averages = {}    # To store {Student: Avg}

    # Tracking variables for school-wide average
    total_marks = 0
    total_count = 0

    # Outer Loop: Iterate through each Class
    for class_name in data:
        highest_avg = 0
        topper = ""

        # Inner Loop: Iterate through each Student in that Class
        for student in data[class_name]:
            marks = data[class_name][student]
            
            # Calculate individual average
            avg = sum(marks) / len(marks)
            student_averages[student] = avg

            # Update school-wide totals for the final school average
            total_marks += sum(marks)
            total_count += len(marks)

            # Logic to identify the topper of the current class
            if avg > highest_avg:
                highest_avg = avg
                topper = student

        # Store the class topper in the required tuple format
        class_toppers[class_name] = (topper, highest_avg)

    school_average = total_marks / total_count if total_count > 0 else 0

    return {
        "class_toppers": class_toppers,
        "school_average": school_average,
        "student_averages": student_averages
    }

# Execute the analysis
result = analyze(school_data)
print(result)


"""
Why Nested Dictionaries are useful for Hierarchical Data:
- They naturally represent real-world "belongs-to" relationships 
   (e.g., Student belongs to a Class, Class belongs to a School).
- To update one class's data without affecting the structure or values of another class.
- Allows to jump straight to a specific category (like "Class B") using a key rather than searching through a long list.
"""
```
